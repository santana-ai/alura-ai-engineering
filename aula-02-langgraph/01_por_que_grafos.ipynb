# [Conceito] Por que grafos?

## Os limites do agente caixa-preta

Um agente pronto decide sozinho, a cada passo, o que fazer a seguir. De fora, só dá para ver a mensagem
que entra e a resposta que sai. O caminho no meio fica invisível: quantos passos ele deu, o que decidiu
em cada um, por que escolheu aquela tool.

Isso custa três coisas: controle (não dá para forçar uma ordem específica de etapas), previsibilidade
(o número de passos varia a cada execução) e depuração (um erro no meio do caminho é difícil de isolar).

## Grafos como modelo mental

Um **grafo** troca a caixa-preta por passos explícitos: **nós** são as etapas de processamento, **arestas**
são as transições entre elas. Um **estado** viaja de nó em nó, e cada nó lê uma parte dele e devolve uma
atualização.

Nada disso é novo: é o mesmo raciocínio de um fluxograma, só que executável.

## O create_agent já é um grafo

O `create_agent` que vimos anteriormente já é, por baixo, um grafo do LangGraph montado e compilado
automaticamente. O ciclo ReAct (responder ou chamar uma tool, repetir até ter o suficiente) é um nó de
modelo e um nó de tools se alternando, com uma aresta condicional decidindo quando parar.

O que muda nesta aula é o nível de controle: em vez de aceitar esse grafo pronto, montamos o nosso.

## Fluxos vs. agentes

Um agente decide dinamicamente a cada passo. Um fluxo (grafo com transições fixas ou condicionais
explícitas) segue um caminho previsível, definido de antemão. Nem toda etapa de uma aplicação precisa da
autonomia de um agente: um fluxo de triagem, por exemplo, ganha mais em previsibilidade e facilidade de
teste do que perderia em flexibilidade.

Essa escolha reflete o quanto de controle a aplicação precisa, e o quanto disso vale abrir mão em troca de
autonomia.
