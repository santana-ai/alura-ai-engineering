# AI Engineering e LangChain

## De simples LLM call para construir uma aplicação

**Usar um LLM**: enviar um texto e receber outro. Resolve tarefas isoladas (resumir, traduzir, classificar). 

Não lembra da conversa, não acessa dados internos, não executa ações, não deixa rastro.

**Uma aplicação com LLM** compõe o modelo com outras peças:
- **prompt**: papel, tom e regras;
- **mensagens**: a conversa (sistema, usuário, assistente);
- **tools**: ações no mundo (consultar agenda, ler banco, chamar API);
- **contexto**: os dados certos na hora certa;
- **fluxo de execução** e **observabilidade**.

Fazer o protótipo é fácil. O difícil é operacionalizar com confiabilidade.

## O que é um agente de IA

Um **agente** é uma abstração em que o próprio LLM decide, a cada passo, o que fazer a seguir: responder ou usar uma **tool**. Repete o ciclo até ter o suficiente para responder.

<img src="resources/react-loop-concept.png" width="50%">

Esse ciclo é o **ReAct**: raciocinar, agir, observar, responder.

## AI Engineering e Context Engineering

**AI Engineering**: a disciplina de construir e operar aplicações com Modelos de IA, principalemnte de linguagem. Cobre o modelo e tudo em volta dele: tools, contexto, fluxo de execução, segurança e observabilidade.

**Context Engineering**: decidir o que entra no prompt a cada passo (instruções, histórico, dados do sistema, conhecimento recuperado). O contexto certo, na hora certa.

<img src="resources/context-eng.png" width="50%">

## A stack LangChain

Peças que trabalham juntas, de baixo para cima:
- **LangGraph**: runtime de orquestração (grafo com estado, execução durável, streaming, human-in-the-loop). A base.
- **LangChain**: framework de agentes, sobre o LangGraph. Abstrações (modelos, mensagens, tools, middleware) e o `create_agent`. O foco deste curso.
- **DeepAgents**: harness pronto, em cima do LangChain, para tarefas longas. Menção do ecossistema, não usado neste curso.
- **LangSmith**: observabilidade (traces e custo dos agentes).
